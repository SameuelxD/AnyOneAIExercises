**Question 3** What is the state with most customers?

```python
from os.path import exists
import csv

# Dict to count the quantity the clients for state
datalog = {}

# Check if the document exists
datafile = "./sample_data/customers.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

# Reader the document CSV
with open(datafile, newline='', encoding='utf-8') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    headers = next(csvreader)

    for row in csvreader:
          state = row[4]
          # Check how many clients there are in each state
          datalog[state] = datalog.get(state, 0) + 1

# Find the state with most clients
state_most_customers = max(datalog, key=datalog.get)  # State with most clients
max_customers = datalog[state_most_customers]  # Clients Number in there is state

print(f"The state with the most customers is {state_most_customers} with {max_customers} customers.")
```

```python
# Question #1: How many unique orders are in the orders.csv file?

from os.path import exists
import csv

# Dict to count the quantity the orders
order_products = {}

# Check if the document exists
datafile = "./sample_data/orders.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

# Reader the document CSV
with open(datafile, newline='', encoding='utf-8') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    headers = next(csvreader)

    for row in csvreader:
      order_id = row[1]
      product_name = row[4]

      if product_name in order_products:
        order_products[product_name].add(order_id)
      else:
        order_products[product_name] = {order_id}

unique_orders = {order for order_set in order_products.values() if len(order_set) == 1 for order in order_set}
unique_orders_count = len(unique_orders)

print(f"The number of unique orders (where products do not repeat in other orders) is: {unique_orders_count}")
```

```python
# Question #2: What is the average number of items per order (rounded to two decimal places)?

from os.path import exists
import csv

# Dict to count the quantity the orders
order_counts = {}

# Check if the document exists
datafile = "./sample_data/orders.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

# Reader the document CSV
with open(datafile, newline='', encoding='utf-8') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    headers = next(csvreader)

    for row in csvreader:
      order_id = row[1]
      order_counts[order_id] = order_counts.get(order_id, 0) + 1

# Calculate the average number of items per order
total_items = sum(order_counts.values())
total_orders = len(order_counts)
average_items = round(total_items / total_orders, 2) if total_orders > 0 else 0

print(f"The average number of items per order is: {average_items}")


```

```python
# Question #3: What is the highest number of items per order?

from os.path import exists
import csv

# Dict to count the quantity the orders
order_counts = {}

# Check if the document exists
datafile = "./sample_data/orders.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

# Reader the document CSV
with open(datafile, newline='', encoding='utf-8') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    headers = next(csvreader)

    for row in csvreader:
      order_id = row[1]
      order_counts[order_id] = order_counts.get(order_id, 0) + 1

# Find the highest number of items
max_items = max(order_counts.values(), default=0)

print(f"The highest number of items in a single order is: {max_items}")
```

```python
# Question #4: What is the number of orders placed in October 2021?

from os.path import exists
import csv
from datetime import datetime

# Dict to count the quantity the orders
order_counts = 0

# Check if the document exists
datafile = "./sample_data/orders.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

# Reader the document CSV
with open(datafile, newline='', encoding='utf-8') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    headers = next(csvreader)
# Count the number of orders placed in October 2021
    for row in csvreader:
      order_date_str = row[2].strip()
      if not order_date_str:
        continue
      try:
        order_date = datetime.strptime(row[2], "%Y-%m-%d %H:%M:%S.%f")
      except ValueError:
        order_date = datetime.strptime(row[2], "%Y-%m-%d %H:%M:%S")
      if order_date.year == 2021 and order_date.month == 10:
        order_counts += 1

print(f"The number of orders placed in October 2021 is: {order_counts}")

```

```python
# Question #5: Which customer spent the most amount of money in 2021?

from os.path import exists
import csv
from datetime import datetime

# Dict to count the quantity the orders
customer_spending = {}

# Check if the document exists
datafile = "./sample_data/orders.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

# Reader the document CSV
with open(datafile, newline='', encoding='utf-8') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    headers = next(csvreader)

# Get the date and filter for the year 2021
    for row in csvreader:
      customer_id = row[0].strip()
      order_date_str = row[2].strip()
      order_total_str = row[3].strip()

      if not order_date_str or not order_total_str:
        continue
      try:
        order_total = float(order_total_str)
      except ValueError:
        continue

      try:
        order_date = datetime.strptime(order_date_str, "%Y-%m-%d %H:%M:%S.%f")
      except ValueError:
        order_date = datetime.strptime(order_date_str, "%Y-%m-%d %H:%M:%S")
      if order_date.year == 2021:
       customer_spending[customer_id] = customer_spending.get(customer_id, 0) + order_total
# Get the customer what most spent in 2021
top_customer = max(customer_spending, key=customer_spending.get)
max_spent = customer_spending[top_customer]
print(f"The customer who spent the most in 2021 is {top_customer} with a total of ${max_spent:.2f}.")

```

```python
# Question #6: Historically, what is the best month for sales?

from os.path import exists
import csv
from datetime import datetime
from collections import defaultdict

# Dictionary to store total sales per (year, month)
monthly_sales = defaultdict(float)

# Check if the file exists
datafile = "./sample_data/orders.csv"
if not exists(datafile):
    raise SystemExit("You should run the first code cell and download the dataset files!")

# Read the CSV file
with open(datafile, newline='', encoding='utf-8') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    headers = next(csvreader)

    for row in csvreader:
        order_date_str = row[2].strip()
        order_total_str = row[3].strip()

        if not order_date_str or not order_total_str:
            continue

        # Convert order total to float
        try:
            order_total = float(order_total_str)
        except ValueError:
            continue

        # Parse the order date
        try:
            order_date = datetime.strptime(order_date_str, "%Y-%m-%d %H:%M:%S.%f")
        except ValueError:
            try:
                order_date = datetime.strptime(order_date_str, "%Y-%m-%d %H:%M:%S")
            except ValueError:
                continue

        # Store total sales per (year, month)
        month_key = (order_date.year, order_date.month)
        monthly_sales[month_key] += order_total

# Find the best month for sales

best_month = max(monthly_sales, key=monthly_sales.get)
best_month_sales = monthly_sales[best_month]  # Total sales in that month
best_month_formatted = f"{best_month[0]}-{best_month[1]:02d}"
print(f"The best month for sales historically is {best_month_formatted} with total sales of ${best_month_sales:.2f}.")
```
